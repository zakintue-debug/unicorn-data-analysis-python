```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import pandas as pd

# Provide the correct file path
file_path = r'C:\Users\kamik\OneDrive\Documents\GitHub\Coursework\UNICORN.csv'

# Load the dataset
df = pd.read_csv(file_path)

# Display the first few rows of the dataset
df.head()
```

```python
#to check the number of columns in the dataset
df.columns
```

```python
#to check the unique value in the dataset column valuation
df["Valuation"].unique()
```

```python
#intialize with the empty list to store values of valuation
Valuations=[]

#iterate over each value of the column valuation
for val in df["Valuation"]:

    #check the any string values
    if isinstance (val, str):

        #replace every string and make it pure numeric to perform some operations
        val=val.replace('$', '').replace('B', '')

        #convert every valuation to float and append it valuation list
        Valuations.append(float(val))

    else:
        #if the value is not string append 0 to valuation list 
        Valuations.append(0)
        
#rename the column     
df['Valuation_USD_Billions'] = Valuations
```

```python
#initiate the variables to store the values of valid valuations and count 
total= 0
count= 0

#iterate over the every row of column "Valuation_USD_Billions" to check in dataframe
for val in df["Valuation_USD_Billions"]:
    
    val=float(val)
    if val>0:
        total += val
        count += 1
# Calculate the average value by dividing the total sum and value count
average_val = total/count

#print the result
print("Average valuation (in billions):",average_val)


    
```

```python
#initiate with zero to store the value of low medium and high
low_count = 0
medium_count = 0
high_count = 0

#use for loop to run on column for checking the categories
for val in df["Valuation_USD_Billions"]:
    if val < 2:
        category = "low"
        low_count += 1
    elif val < 10:
        category = "medium"
        medium_count += 1
    else:
        category = "high"
        high_count += 1

# Display the counts of each category
print("Low Valuation Count:", low_count)
print("Medium Valuation Count:", medium_count)
print("High Valuation Count:", high_count)
```

```python
#calculate the threshold which is double the valuation
threshold = average_val*2

#print the result of that columm
print("high valuation company :")

#check each row of dataframe
for i in range (len(df)):
    
    #check the threshold 
    if df["Valuation_USD_Billions"][i] > threshold:

        #print the compaines which complete on that criteria
        print(df ["Company"][i], "--", df["Valuation_USD_Billions"][i],"B")
```

```python
#start with zero so store the value 
i = 0

#run the loop to check in dataframe
while i < len(df):

    #use if so to check which company reach the level of 50B first 
    if df["Valuation_USD_Billions"][i] > 50:

        #print that result 
        print("First company with valuation above 50B:", df["Company"][i])

       #else break that 
        break
    #add +1 in every time so check every row of dataframe    
    i+=1
```

```python
#initialize with empty dictionary to store the values of each industry of the companies
industry_count={}

#iterate over every column of the dataframe
for  industry in df ["Industry"]:

    #check if the industry already exist
    if industry in industry_count:
        #if yes increase 1 into it every time 
        industry_count[industry] +=1
    else :
        #if do not add it in dictionary with count of 1 only 
        industry_count[industry] =1

# Convert the dictionary into a list of tuples     
industry_list = list(industry_count.items())

#print the results
print(industry_list)
```

```python
#initialize with the zero values to store values of that
before_2012 = 0
after_2012 = 0

#run for lopp to check the year founded in datafram column year founded
for year in df["Year Founded"]:

    #if the year is less then 2012
    if year  < 2012 :
        #assign that into before 2012
        before_2012 += 1
    else:
        #assign that into after 2012
        after_2012 += 1
#print the results of that before and after
print("Companies founded before 2012:",before_2012)
print("Companies founded after 2012:",after_2012)
```

```python
import pandas as pd
import numpy as np
import requests
import matplotlib.pyplot as plt
```

```python
# Function to load the dataset
def load_data(file_path):

    # Use pandas to read the CSV file from the given file directory.
    data = pd.read_csv(file_path)

    #return the dataframe
    return data
```

```python
# Create a function to determine a column's average
def calculate_avg(column):

    # To determine the average, add up all of the values in the column and divide by the total number of items.
    return sum(column)/len(column)

# Use the function "calculate_avg" on the DataFrame "df"'s "Year Founded" column.    
avg_year_founded = calculate_avg(df["Year Founded"])

# Print the "Year Founded" column's computed average.
print(f"Average year founded: {avg_year_founded}")
```

```python
# check high values
def count_of_high_values(column,threshold):
    count=0
    for value in column:
        if value>threshold:
            count+=1
    return count

# Set the threshold value
threshold_value = 20 

# Apply the function to the Valuation_USD_Billions
high_valuation_count = count_of_high_values(df['Valuation_USD_Billions'], threshold_value)

# Print the result
print(f"Companies with the greater valuation then the {threshold_value} billion USD are: {high_valuation_count}")
```

```python
# Create a function that counts the distinct values in a column.
def count_unique_values(Company,):
    # Return the set's length after converting the column to a set.
    return len(set(Company))

# Extract the 'Industry' column from the 'df' DataFrame.
industry_column = df['Industry']

# Use the 'count_unique_values' function on the 'Industry' column, then report the outcome.
print(f"The unique industry in the dataset are: ",count_unique_values(industry_column))
```

```python
# To count frequencies, import the 'Counter' class from the 'collections' module.
from collections import Counter

# Make a function that counts how many times each unique number appears in a column.
def frequency_count(column):

    #Use Counter to find out how many times each number appears in the column and turn it into a dictionary.
    return dict(Counter(column))

# Get the 'Industry' column from the DataFrame 'df'
industry_column = df['Industry']

#print the result of frequncey_count
print(frequency_count(industry_column))
```

```python
# Make a function that counts the number of non-null items in a certain column.
def count_non_null_values(column):

    # To get rid of None, empty strings, "NaN," and "nan" numbers, use a list comprehension.
    return len([x for x in column if x not in (None, '', 'NaN', 'nan')])

# Get the DataFrame "df"'s "Industry" column    
industry_column = df['Industry']


# Count non-null values
print(count_non_null_values(industry_column))
```

```python
def most_frequent_value(column):
    return Counter(column).most_common(1)[0][0]

industry_column = df['Industry']
# Most frequent occuering 
print(most_frequent_value(industry_column))
```

```python
def top_n_frequent_values(column, n):
    return [item[0] for item in Counter(column).most_common(n)]

industry_column = df['Industry']

# Get top 3 most frequent values
print(top_n_frequent_values(industry_column, 3))
```

```python
# From the DataFrame 'df', select the 'Company' and 'Valuation_USD_Billions' columns 'Valuation_USD_Billions' in 
#decreasing order to obtain the top 10 firms.

top_10_unicorn_companies = df[['Company', 'Valuation_USD_Billions']].sort_values(by='Valuation_USD_Billions', ascending=False).head(10)
```

```python
import matplotlib.pyplot as plt

# Create a pie chart
plt.figure(figsize=(8,8))

# Calculate the total valuation to represent each company's percentage
total_valuation = top_10_unicorn_companies['Valuation_USD_Billions'].sum()

# Plot the pie chart using the companies and their respective valuations
plt.pie(top_10_unicorn_companies['Valuation_USD_Billions'], 
        labels=top_10_unicorn_companies['Company'], 
        autopct=lambda p: f'{p:.1f}% ({p*total_valuation/100:.1f}B)', 
        startangle=140, 
        colors=['#FF9999','#66B3FF','#99FF99','#FFCC99','#FFD700','#DA70D6','#32CD32','#8A2BE2','#FF6347','#3CB371'])

# Add title
plt.title('Top 10 Unicorns by Valuation', fontsize=16)

# Ensure the pie is drawn as a circle
plt.axis('equal')

# Display the plot
plt.tight_layout()
plt.show()
```

```python
# Count the number of companies per continent
continent_counts = df['Continent'].value_counts()

# Create a bar chart
plt.figure(figsize=(10,6))
plt.bar(continent_counts.index, continent_counts, color=['#FF9999','#66B3FF','#99FF99','#FFCC99'])

# Add labels and title
plt.title('Unicorn Companies Distribution by Continent', fontsize=16)
plt.xlabel('Continent', fontsize=12)
plt.ylabel('Number of Companies', fontsize=12)

# Rotate the x-axis labels for better readability
plt.xticks(rotation=45)

# Add a grid for easier comparison
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Display the bar chart
plt.tight_layout()  # Ensures everything fits well in the plot
plt.show()


```

```python
import tkinter as tk
from tkinter import ttk, messagebox
```

```python
# Create the main window
root = tk.Tk()
#Set the window welcome page title 
root.title("Welcome Page")
#Use the window size 800X600 pixel
root.geometry("800x600")
```

```python
# Function to move to the next screen after displaying the welcome message
def go_to_main_screen():
    welcome_frame.pack_forget()  # Hide the welcome page
    main_screen()  # Show the main screen with options
```

```python
# Welcome page with NTU ID and a brief message
welcome_frame = tk.Frame(root)
welcome_frame.pack(fill=tk.BOTH, expand=True)

welcome_label = tk.Label(welcome_frame, text="Welcome to Unicorn Companies Data", font=("Helvetica", 20))
welcome_label.pack(pady=20)

ntu_id_label = tk.Label(welcome_frame, text="NTU ID: N1420010", font=("Helvetica", 14))
ntu_id_label.pack(pady=20)

# Button to proceed to the main menu
next_button = tk.Button(welcome_frame, text="Proceed", font=("Helvetica", 14), command=go_to_main_screen)
next_button.pack(pady=20)
```

```python
# Main screen with options (e.g., View Data, Search Companies)
def main_screen():
    main_frame = tk.Frame(root)
    main_frame.pack(fill=tk.BOTH, expand=True)

    options_label = tk.Label(main_frame, text="Choose an option", font=("Helvetica", 16))
    options_label.pack(pady=20)

    # Button to show the data in Treeview
    show_data_button = tk.Button(main_frame, text="View Unicorn Companies", font=("Helvetica", 14), command=view_data)
    show_data_button.pack(pady=10)

    # Button to show search functionality
    search_button = tk.Button(main_frame, text="Search Unicorn Companies", font=("Helvetica", 14), command=show_search_page)
    search_button.pack(pady=10)
```

```python
# Function to display the unicorn companies' data
def view_data():
    root.geometry("1000x600")  # Resize window for data view

    main_frame = tk.Frame(root)
    main_frame.pack(fill=tk.BOTH, expand=True)

    # Create Treeview widget to display the data
    columns = ['Company', 'Valuation', 'Date Joined', 'Industry', 'City', 'Country', 'Continent', 'Year Founded', 'Funding', 'Select Investors']
    tree = ttk.Treeview(main_frame, columns=columns, show="headings")

    for col in columns:
        tree.heading(col, text=col)
        tree.column(col, width=120)

    # Insert data into the treeview
    for index, row in df.iterrows():
        tree.insert("", "end", values=list(row))

    tree.pack(fill=tk.BOTH, expand=True)

    # Add a button to go back to the options screen
    back_button = tk.Button(main_frame, text="Back to Options", font=("Helvetica", 14), command=go_back)
    back_button.pack(pady=10)
```

```python

# Function to go back to the main options screen
def go_back():
    for widget in root.winfo_children():
        widget.pack_forget()
    main_screen()

# Search Page
def show_search_page():
    root.geometry("600x400")  # Resize window for search
    search_frame = tk.Frame(root)
    search_frame.pack(fill=tk.BOTH, expand=True)

    search_label = tk.Label(search_frame, text="Enter Company Name to Search", font=("Helvetica", 14))
    search_label.pack(pady=20)

    # Search bar
    search_entry = tk.Entry(search_frame, font=("Helvetica", 14))
    search_entry.pack(pady=10)

    # Function to search and display results
    def search_companies():
        query = search_entry.get().lower()
        for row in tree.get_children():
            tree.delete(row)

        # Filter the DataFrame based on the search query and reinsert data
        filtered_df = df[df['Company'].str.lower().str.contains(query)]
        for index, row in filtered_df.iterrows():
            tree.insert("", "end", values=list(row))

    # Create a search button
    search_button = tk.Button(search_frame, text="Search", font=("Helvetica", 14), command=search_companies)
    search_button.pack(pady=10)

    # Treeview to display search results
    columns = ['Company', 'Valuation', 'Date Joined', 'Industry', 'City', 'Country', 'Continent', 'Year Founded', 'Funding', 'Select Investors']
    tree = ttk.Treeview(search_frame, columns=columns, show="headings")
    
    for col in columns:
        tree.heading(col, text=col)
        tree.column(col, width=120)

    tree.pack(fill=tk.BOTH, expand=True)

    # Add a button to go back to the options screen
    back_button = tk.Button(search_frame, text="Back to Options", font=("Helvetica", 14), command=go_back)
    back_button.pack(pady=10)

# Start the welcome page
root.mainloop()
```

It also emphasizes the rapid growth and valuation increase of unicorns over the past few decades, particularly since 2018, as well as the significant capital required to reach these billion-dollar valuations. Geographic and industry-specific patterns emerge, showing the concentration of unicorns in specific cities and countries, notably in tech hubs like the San Francisco Bay Area and Beijing. Through interactive dashboards and visualizations, these insights are presented in a clear and engaging manner, offering valuable information for entrepreneurs, investors, and policymakers to navigate the dynamic unicorn landscape.
