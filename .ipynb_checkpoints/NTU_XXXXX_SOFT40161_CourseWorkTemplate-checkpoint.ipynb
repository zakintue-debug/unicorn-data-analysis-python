```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import pandas as pd

# Provide the correct file path
file_path = r'C:\Users\kamik\OneDrive\Documents\GitHub\Coursework\UNICORN.csv'

# Load the dataset
df = pd.read_csv(file_path)

# Display the first few rows of the dataset
df.head()
```

```python
df.columns
```

```python
df["Valuation"].unique()
```

```python
Valuations=[]

for val in df["Valuation"]:

    if isinstance (val, str):

        val=val.replace('$', '').replace('B', '')

        Valuations.append(float(val))

    else:
        Valuations.append(0)
    
df['Valuation_USD_Billions'] = Valuations
```

```python
total= 0
count= 0

for val in df["Valuation_USD_Billions"]:
    val=float(val)
    if val>0:
        total += val
        count += 1

average_val = total/count
print("Average valuation (in billions):",average_val)


    
```

```python
low_count = 0
medium_count = 0
high_count = 0

for val in df["Valuation_USD_Billions"]:
    if val < 2:
        category = "low"
        low_count += 1
    elif val < 10:
        category = "medium"
        medium_count += 1
    else:
        category = "high"
        high_count += 1

# Display the counts of each category
print("Low Valuation Count:", low_count)
print("Medium Valuation Count:", medium_count)
print("High Valuation Count:", high_count)
```

```python
threshold = average_val*2

print("high valuation company :")
for i in range (len(df)):
    if df["Valuation_USD_Billions"][i] > threshold:
        print(df ["Company"][i], "--", df["Valuation_USD_Billions"][i],"B")
```

```python
i = 0
while i < len(df):
    if df["Valuation_USD_Billions"][i] > 50:
        print("First company with valuation above 50B:", df["Company"][i])
        break
    i+=1
```

```python
industry_count={}

for  industry in df ["Industry"]:
    if industry in industry_count:
        industry_count[industry] +=1
    else :
        industry_count[industry] =1
industry_list = list(industry_count.items())
print(industry_list)
```

```python
before_2012 = 0
after_2012 = 0

for year in df["Year Founded"]:
    if year  < 2012 :
        before_2012 += 1
    else:
        after_2012 += 1

print("Companies founded before 2012:",before_2012)
print("Companies founded after 2012:",after_2012)
```

```python
# create a load_data function using pandas library read_csv() method to import data from the file in which is already defined
# in the file_path in section 1. And return the dataset as the pandas dataframe.
```

```python
# Function to load the dataset
def load_data(file_path):
    data = pd.read_csv(file_path)
    return data
```

```python
create calculate_avg function to get the average of any column in our case we  use "year founded" to know what the average year of founding these
unicorn industires.
```

```python
# Function to calculate an average
def calculate_avg(column):
    return sum(column)/len(column)
avg_year_founded = calculate_avg(df["Year Founded"])

print(f"Average year founded: {avg_year_founded}")
```

```python
Create "count_of_high_values" function so we can know the highest values unicorn companies greater then the given threshold. 
Threshold is set 20 billion and applied on "Valuation_USD_Billions" column. Give us the number of companies of valuation exceed 20 billion.
```

```python
# check high values
def count_of_high_values(column,threshold):
    count=0
    for value in column:
        if value>threshold:
            count+=1
    return count

# Set the threshold value
threshold_value = 20 

# Apply the function to the Valuation_USD_Billions
high_valuation_count = count_of_high_values(df['Valuation_USD_Billions'], threshold_value)

# Print the result
print(f"Companies with the greater valuation then the {threshold_value} billion USD are: {high_valuation_count}")
```

```python
def count_unique_values(Company,):
    return len(set(Company))
industry_column = df['Industry']
print(f"The unique industry in the dataset are: ",count_unique_values(industry_column))
```

```python
from collections import Counter

def frequency_count(column):
    return dict(Counter(column))
    
industry_column = df['Industry']

print(frequency_count(industry_column))
```

```python
def count_non_null_values(column):
    return len([x for x in column if x not in (None, '', 'NaN', 'nan')])
industry_column = df['Industry']
# Count non-null values
print(count_non_null_values(industry_column))
```

```python
def most_frequent_value(column):
    return Counter(column).most_common(1)[0][0]

industry_column = df['Industry']
# Most frequent occuering 
print(most_frequent_value(industry_column))
```

```python
def values_appearing_once(column):
    
    return [item for item, count in Counter(column).items() if count == 1]

industry_column = df['Industry']
# Get values that appear only once
print(values_appearing_once(industry_column))
```

```python
def top_n_frequent_values(column, n):
    return [item[0] for item in Counter(column).most_common(n)]

industry_column = df['Industry']

# Get top 3 most frequent values
print(top_n_frequent_values(industry_column, 3))
```

```python
import tkinter as tk
from tkinter import ttk, messagebox
```

```python
# Create the main window
root = tk.Tk()
root.title("Welcome Page")
root.geometry("800x600")
```

```python
# Function to move to the next screen after displaying the welcome message
def go_to_main_screen():
    welcome_frame.pack_forget()  # Hide the welcome page
    main_screen()  # Show the main screen with options
```

```python
# Welcome page with NTU ID and a brief message
welcome_frame = tk.Frame(root)
welcome_frame.pack(fill=tk.BOTH, expand=True)

welcome_label = tk.Label(welcome_frame, text="Welcome to Unicorn Companies Data", font=("Helvetica", 20))
welcome_label.pack(pady=20)

ntu_id_label = tk.Label(welcome_frame, text="NTU ID: N1420010", font=("Helvetica", 14))
ntu_id_label.pack(pady=20)

# Button to proceed to the main menu
next_button = tk.Button(welcome_frame, text="Proceed", font=("Helvetica", 14), command=go_to_main_screen)
next_button.pack(pady=20)
```

```python
# Main screen with options (e.g., View Data, Search Companies)
def main_screen():
    main_frame = tk.Frame(root)
    main_frame.pack(fill=tk.BOTH, expand=True)

    options_label = tk.Label(main_frame, text="Choose an option", font=("Helvetica", 16))
    options_label.pack(pady=20)

    # Button to show the data in Treeview
    show_data_button = tk.Button(main_frame, text="View Unicorn Companies", font=("Helvetica", 14), command=view_data)
    show_data_button.pack(pady=10)

    # Button to show search functionality
    search_button = tk.Button(main_frame, text="Search Unicorn Companies", font=("Helvetica", 14), command=show_search_page)
    search_button.pack(pady=10)
```

```python
# Function to display the unicorn companies' data
def view_data():
    root.geometry("1000x600")  # Resize window for data view

    main_frame = tk.Frame(root)
    main_frame.pack(fill=tk.BOTH, expand=True)

    # Create Treeview widget to display the data
    columns = ['Company', 'Valuation', 'Date Joined', 'Industry', 'City', 'Country', 'Continent', 'Year Founded', 'Funding', 'Select Investors']
    tree = ttk.Treeview(main_frame, columns=columns, show="headings")

    for col in columns:
        tree.heading(col, text=col)
        tree.column(col, width=120)

    # Insert data into the treeview
    for index, row in df.iterrows():
        tree.insert("", "end", values=list(row))

    tree.pack(fill=tk.BOTH, expand=True)

    # Add a button to go back to the options screen
    back_button = tk.Button(main_frame, text="Back to Options", font=("Helvetica", 14), command=go_back)
    back_button.pack(pady=10)
```

```python

# Function to go back to the main options screen
def go_back():
    for widget in root.winfo_children():
        widget.pack_forget()
    main_screen()

# Search Page
def show_search_page():
    root.geometry("600x400")  # Resize window for search
    search_frame = tk.Frame(root)
    search_frame.pack(fill=tk.BOTH, expand=True)

    search_label = tk.Label(search_frame, text="Enter Company Name to Search", font=("Helvetica", 14))
    search_label.pack(pady=20)

    # Search bar
    search_entry = tk.Entry(search_frame, font=("Helvetica", 14))
    search_entry.pack(pady=10)

    # Function to search and display results
    def search_companies():
        query = search_entry.get().lower()
        for row in tree.get_children():
            tree.delete(row)

        # Filter the DataFrame based on the search query and reinsert data
        filtered_df = df[df['Company'].str.lower().str.contains(query)]
        for index, row in filtered_df.iterrows():
            tree.insert("", "end", values=list(row))

    # Create a search button
    search_button = tk.Button(search_frame, text="Search", font=("Helvetica", 14), command=search_companies)
    search_button.pack(pady=10)

    # Treeview to display search results
    columns = ['Company', 'Valuation', 'Date Joined', 'Industry', 'City', 'Country', 'Continent', 'Year Founded', 'Funding', 'Select Investors']
    tree = ttk.Treeview(search_frame, columns=columns, show="headings")
    
    for col in columns:
        tree.heading(col, text=col)
        tree.column(col, width=120)

    tree.pack(fill=tk.BOTH, expand=True)

    # Add a button to go back to the options screen
    back_button = tk.Button(search_frame, text="Back to Options", font=("Helvetica", 14), command=go_back)
    back_button.pack(pady=10)

# Start the welcome page
root.mainloop()
```
